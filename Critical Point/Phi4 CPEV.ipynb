```julia
using Plots
using Plots.PlotMeasures
using HDF5
```

```julia
N = 15 #Number of lattice sites per dimension
d = 1 #Number of spatial dimensions
Dim = 10 #Truncated local Hilbert space dimension
a = 1.0 #Lattice spacing

n_0 = round(Int, ((N-1)/2)) #Index of the point at the center of the lattice (0-based indexing).

m0 = 1.0 #Basis frequency
l = 0.1 #phi^4 coupling strength

m_i = 0.01 #first renormalized mass
m_f = 0.1 #last renormalized mass
len = 20
m_range = 10 .^ range(log10(m_i), log10(m_f), length=len);
```

```julia
plt = scatter(guidefont=font(12),
    xlabel="\${\\lambda}/{m_{R}^{2}}\$",
    ylabel="\$\\langle\\phi\\rangle\$")

data = h5open("Z:/Energy Correlator/1d_EEC_data/CPEV,N=$N,a=$a,l=$l,m_i=$m_i,m_f=$m_f,len=$len", "r")
VEVs = data["VEVs"][1:end]
close(data)
scatter!((l ./ m_range.^2), real(VEVs))

plt
#savefig("Plots/Energy Conservation, N=$N, m=$mass, lambda=1e-1, truncated HS dim=$Dim")
```
