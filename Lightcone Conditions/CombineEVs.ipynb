```julia
using HDF5
N = 15 #Number of lattice sites per dimension
d = 1 #Number of spatial dimensions
Dim = 10 #Truncated local Hilbert space dimension
a = 1.0 #Lattice spacing

n_0 = round(Int, ((N-1)/2)) #Index of the point at the center of the lattice (0-based indexing).

mass = 1.0
m0 = 1.0 #Basis frequency
l = 1.0 #phi^4 coupling strength

r=5
t_ranges = [[0.0, 5.0], [5.2, 10]]
dt=0.2

EVs = ComplexF64[]
for i in 1:length(t_ranges)
    EEC_data = h5open("Z:/Energy Correlator/1d_EEC_data/EFEV,N=$N,a=$a,l=$l,m=$mass,r=$r,ti=$(t_ranges[i][1]),tf=$(t_ranges[i][2]),dt=$dt", "r")
    EV = EEC_data["Eflux"][1:end]
    close(EEC_data)
    append!(EVs, EV)
end

data = h5open("Z:/Energy Correlator/1d_EEC_data/EFEV,N=$N,a=$a,l=$l,m=$mass,r=$r,ti=$(t_ranges[1][1]),tf=$(t_ranges[end][2]),dt=$dt", "w")
data["Eflux"] = EVs
close(data)
```
